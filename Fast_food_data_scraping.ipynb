## Generate Grid of Subarea Centers

### Subtask:
Calculate the latitude and longitude for the centers of multiple sub-circles, ensuring they collectively cover the specified total radius area around the central point (center_lat, center_long), with each sub-circle having subarea_radius. This will create a grid of points to be used for subsequent API calls.


**Reasoning**:
First, I'll import the math module, which is essential for trigonometric calculations needed to convert radius to degrees and determine accurate offsets for longitude based on latitude. Then, I will initialize some constants and calculate the number of steps and angular offsets based on the provided instructions to prepare for generating the grid of subarea centers.



```python
## VERIFIED
import math

# Convert total_radius from meters to kilometers
total_radius_km = total_radius / 1000.0

# Number of steps required to cover the total_radius based on the subarea_radius
# Each subarea_radius is in meters, so convert to km for consistency with total_radius_km
subarea_radius_km = subarea_radius / 1000.0
num_steps_from_center = int(total_radius_km / subarea_radius_km) + 1 # +1 to ensure coverage
## --> num_steps_from_center calculates how many sub-areas you need to 'step' out from the center to reach
## the edge of your total_radius (27 km). Imagine dividing the 27 km total radius into segments,
## each the size of your subarea_radius_km. The + 1 ensures you cover the entire distance,
## including the center and the farthest edge.

# Constants for degree-to-km conversion (approximate for earth)
KM_PER_DEGREE_LATITUDE = 111.32 # km per degree of latitude

# let's calculate maximum angular offset from the center
max_lat_offset_degrees = total_radius_km / KM_PER_DEGREE_LATITUDE
## --> max_lat_offset_degrees calculates the maximum number of degrees of latitude we need to move
## from our center_lat to cover the total_radius_km in the north-south direction.
max_lon_offset_degrees = total_radius_km / (KM_PER_DEGREE_LATITUDE * math.cos(math.radians(center_lat)))
## --> max_lon_offset_degrees calculates the maximum number of degrees of longitude we need to move
## from our center_long to cover the total_radius_km in the east-west direction.
## --> math.cos(math.radians(center_lat)) : the length of a
## degree of longitude changes depending on the latitude (it's widest at the equator
## and shrinks towards the poles). This correction accounts for that spherical geometry.


# let's calculate degree increment for each step
lat_step_degrees = max_lat_offset_degrees / num_steps_from_center
## --> lat_step_degrees = max_lat_offset_degrees / num_steps_from_center divides
## the total latitude offset by the number of steps. It tells you how many degrees
## you need to move in latitude for each grid point, effective
lon_step_degrees = max_lon_offset_degrees / num_steps_from_center
## --> lon_step_degrees = max_lon_offset_degrees / num_steps_from_center divides
## the total longitude offset by the number of steps. It tells you how many degrees
## you need to move in longitude for each grid point, defining the horizontal spacing
## between the columns of your grid.

# let's create an empty list for grid points
grid_points = []

# Generate grid points
for i in range(-num_steps_from_center, num_steps_from_center + 1):
## --> +1 here because python takes [-num_steps_from_center, num_steps_from_center[
    current_lat = center_lat + i * lat_step_degrees
    for j in range(-num_steps_from_center, num_steps_from_center + 1):
        current_lon = center_long + j * lon_step_degrees
        grid_points.append({'latitude': current_lat, 'longitude': current_lon})

print(f"Generated {len(grid_points)} grid points.")
print(f"Example grid point: {grid_points[0]}")
```

## Collect Fast Food Data from Subareas

### Subtask:
Iterate through each generated subarea center. For each center, make an API request to the Google Places API to find fast food restaurants within the defined subarea_radius. Collect all responses into a list.


```python
import time
import requests
import json

# Initialize a list to store all API responses
all_api = []

# Define the API key (using the API key provided in previous cells)
API_KEY = "AIzaSyC2RleANtod74mwmXYiH0q1_52EUmlvLGg"

# Headers for the API request
headers = {
    "Content-Type": "application/json",
    "X-Goog-Api-Key": API_KEY,
    "X-Goog-FieldMask": "places.displayName,places.location.latitude,places.location.longitude"
}

# Base URL for the Places API searchNearby endpoint
PLACES_API_URL = "https://places.googleapis.com/v1/places:searchNearby"

print(f"Starting to collect fast food data from {len(grid_points)} subareas.")
print("This process might take a while and consume API quotas. Please be mindful of your Google Cloud billing.")
print("Each API call retrieves up to 20 places within the subarea_radius.")

# let's iterate through each generated grid point (from the grid_points list)
for i, point in enumerate(grid_points):
    current_lat = point['latitude']
    ## ---> point refers to an indivisual dictionary within the grid_points list
    current_lon = point['longitude']

    payload = {
        "includedTypes": ["fast_food_restaurant"],
        "locationRestriction": {
            "circle": {
                "center": {
                    "latitude": current_lat,
                    "longitude": current_lon
                },
                "radius": subarea_radius  # Use the defined subarea_radius, ie 1.5km
            }
        }
    }

## ==> TRY and EXCEPT blocs <==:
    try:
    ## --> try marks the beginning of a 'try' block. Code inside this block is executed,
    ## and if an error (exception) occurs during its execution, the program doesn't crash
    ## but instead jumps to the corresponding except block.
        response = requests.post(PLACES_API_URL, json=payload, headers=headers)
        ## --> This line sends a POST request to the Google Places API.
        ## --> json=payload sends the payload dictionary (which contains information
        ## about the search, like the location and radius) as a JSON-formatted body in the request.
        response.raise_for_status()
        ## Deals with error messages. Raises an HTTPError for bad responses (4xx or 5xx)
        ## (e.g. 404 Error Found, 500 Internal Error). If the request was successful (2xx status code),
        ## it does nothing and the code continues.
        response_json = response.json()
        if 'places' in response_json and response_json['places']:
            all_api.append(response_json)
            ## --> deals with empty spaces resulting from the 2xx messages (meaning the place is actually empty)

        ## --> This just gives me feedback on how my analysis is going. Print progress every 100 subareas or at the end
        if (i + 1) % 100 == 0 or i == len(grid_points) - 1:
            print(f"Processed {i + 1}/{len(grid_points)} subareas. Collected {len(all_api)} non-empty responses so far.")

    ## --> AI Help to deal with specific error messages (like reach of quota/minute).
    ## Not necessary, just additional personah help to understand better the nature of an error
    except requests.exceptions.HTTPError as errh:
        print(f"HTTP Error for point {i} ({current_lat}, {current_lon}): {errh}")
        if response.status_code == 429:
            print("Rate limit hit. Waiting for 60 seconds before continuing...")
            time.sleep(60)  # Wait for 1 minute
        elif response.status_code == 403:
            print("Forbidden (403) error. Check your API key and ensure the Places API is enabled for your project.")
            break # Stop if API key is likely invalid or not enabled
        elif response.status_code == 400:
            print(f"Bad Request (400) error. Check payload format. Response: {response.text}")
    except requests.exceptions.ConnectionError as errc:
        print(f"Error Connecting for point {i} ({current_lat}, {current_lon}): {errc}")
    except requests.exceptions.Timeout as errt:
        print(f"Timeout Error for point {i} ({current_lat}, {current_lon}): {errt}")
    except requests.exceptions.RequestException as err:
        print(f"Something went wrong with the request for point {i} ({current_lat}, {current_lon}): {err}")
    except json.JSONDecodeError:
        print(f"Failed to decode JSON from response for point {i} ({current_lat}, {current_lon}). Response text: {response.text}")

    # Introduce a small delay to prevent hitting API rate limits.
    # Google Places API has rate limits (e.g., 100 requests per second for free tier).
    # A 0.1 second delay for 1521 requests means approx 2.5 minutes total delay.
    time.sleep(0.1)

print(f"\nFinished collecting data. Total non-empty API responses collected: {len(all_api)}")
```

## Process and Deduplicate Data

### Subtask:
Extract restaurant names, latitudes, and longitudes from all collected API responses. Combine them into a single pandas DataFrame and remove any duplicate entries, ensuring each unique restaurant appears only once.

```python
import pandas as pd

all_places_data = []

# Iterate through each API response collected
for response_json in all_api:
    # Get the 'places' list from each response, if it exists
    places_in_response = response_json.get('places', [])

    # Iterate through each place in the current response
    for place in places_in_response:
        name = place.get('displayName', {}).get('text')
        location = place.get('location', {})
        lat = location.get('latitude')
        lng = location.get('longitude')

        # Only add to list if all required data is present
        if name and lat is not None and lng is not None:
            all_places_data.append({"Name": name, "Latitude": lat, "Longitude": lng})

# Create a DataFrame from the collected data
df_restaurants = pd.DataFrame(all_places_data)

# Remove duplicate entries based on Name, Latitude, and Longitude
df_restaurants.drop_duplicates(subset=['Name', 'Latitude', 'Longitude'], inplace=True)

print(f"Total unique fast food restaurants found: {len(df_restaurants)}")
display(df_restaurants.head())
```
