## **Count of outlets**

```python
import pandas as pd
from google.colab import files
import io

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

# Specify 'header=2' to indicate that the actual column names are in the 3rd row (0-indexed).
# However, sometimes pandas might still create 'Unnamed' columns if the first column of the header row is empty.
ward_data = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name="Ward Data", header=2)

# The actual column names are currently in the first row of the DataFrame (index 0)
# after reading with header=2, but pandas created 'Unnamed' columns. We need to manually set them.
# Clean up column names by stripping whitespace and converting to string.
new_columns = [str(col).strip() if pd.notna(col) else f'Unnamed: {i}' for i, col in enumerate(ward_data.iloc[0])]

# Assign these cleaned names as the DataFrame's columns.
ward_data.columns = new_columns

# Drop the first row of the DataFrame, as it was just used for column names.
ward_data = ward_data.iloc[1:].reset_index(drop=True)

# Remove any completely empty columns that might have been created during the read process
ward_data = ward_data.dropna(axis=1, how='all')

ward_data.head()

london_ward_data = ward_data[ward_data["PHE centre"] =="London"]
```

```python
london_ward_data.sort_values(by='Count of outlets', ascending=False).head(20)
```

```python
london_ward_data.sort_values(by='Count of outlets', ascending=True).head(20)
```

```python
outlet_data = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name="Local Authority Data", header=2)

# The actual column names are currently in the first row of the DataFrame (index 0)
# after reading with header=2, but pandas created 'Unnamed' columns. We need to manually set them.
# Clean up column names by stripping whitespace and converting to string.
new2_columns = [str(col).strip() if pd.notna(col) else f'Unnamed: {i}' for i, col in enumerate(outlet_data.iloc[0])]

# Assign these cleaned names as the DataFrame's columns.
outlet_data.columns = new2_columns

# Drop the first row of the DataFrame, as it was just used for column names.
outlet_data = outlet_data.iloc[1:].reset_index(drop=True)

# Remove any completely empty columns that might have been created during the read process
outlet_data = outlet_data.dropna(axis=1, how='all')

# Convert only the numeric columns to numeric, coercing errors to NaN
numeric_cols = ['Count of outlets', 'Rate per 100,000 population']
for col in numeric_cols:
    if col in outlet_data.columns:
        outlet_data[col] = pd.to_numeric(outlet_data[col], errors='coerce')

# Corrected column name from 'PHE centre' to 'PHE Centre' (as per new2_columns assignment)
london_outlet_data = outlet_data[outlet_data["PHE Centre"] =="London"]

london_outlet_data.sort_values(by='Rate per 100,000 population', ascending=False).head()
```

```python
london_outlet_data.sort_values(by='Rate per 100,000 population', ascending=True).head()
```

## **Childhood Obesity**

```python
import pandas as pd
from google.colab import files
import io

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
obesity_data = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name="2011-12_2013-14")
```

```python
import pandas as pd
import io

# To fix the KeyError, we need to ensure the 'obesity_data' DataFrame has correct column names.
# This involves re-loading the data with appropriate header handling, similar to how 'ward_data' was processed.
# Assuming 'uploaded' and 'file_name' are available from the previous cell's execution in the kernel state.

# Re-read the Excel file, specifying 'header=2' to potentially capture the row with actual column names.
# The sheet name '2011-12_2013-14' was determined from a previous successful load.
# Using header=2 often results in the *next* row (iloc[0]) having the actual desired headers.
initial_obesity_data = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name="2011-12_2013-14", header=2)

# The actual column names are likely in the first row of this 'initial_obesity_data' DataFrame (index 0).
# Clean up column names by stripping whitespace and converting to string.
new_obesity_columns = [str(col).strip() if pd.notna(col) else f'Unnamed: {i}' for i, col in enumerate(initial_obesity_data.iloc[0])]

# Assign these cleaned names as the DataFrame's columns.
obesity_data = initial_obesity_data.copy() # Make a copy to avoid SettingWithCopyWarning
obesity_data.columns = new_obesity_columns

# Drop the first row of the DataFrame, as it was just used for column names.
obesity_data = obesity_data.iloc[1:].reset_index(drop=True)

# Remove any completely empty columns that might have been created during the read process
obesity_data = obesity_data.dropna(axis=1, how='all')

# Find the column that contains "obese" or "% obese" in its name.
obese_col = None
for col in obesity_data.columns:
    if "obese" in str(col).lower():
        obese_col = col
        break

if obese_col is None:
    # If the direct search still fails, you'll need to manually identify the correct column.
    # Please inspect the 'DEBUG' output above and your Excel file to find the column containing obesity percentage.
    # For example, if it's 'Unnamed: 10', you would set: obese_col = 'Unnamed: 10'
    print(f"DEBUG: Available columns in obesity_data: {obesity_data.columns.tolist()}")
    # Manual intervention needed: replace 'YOUR_OBESE_COLUMN_NAME' with the correct column name from the DEBUG list.
    # Based on the user's input, Column N which is 'Unnamed: 13' is the correct one.
    obese_col = 'Unnamed: 13' # Set to 'Unnamed: 13' as per user's clarification
    # raise ValueError("Could not find a column related to 'obese' percentage. Please check the Excel sheet for correct column name from the DEBUG list above, and manually set 'obese_col' in the code.")

# Rename the identified column to 'Year6_%_obese'
obesity_data = obesity_data.rename(columns={obese_col: "Year6_%_obese"})

# Ensure the column is numeric. Coerce errors to NaN for non-numeric values.
obesity_data["Year6_%_obese"] = pd.to_numeric(obesity_data["Year6_%_obese"], errors='coerce')

# Convert to percentage (multiply by 100 if it's a decimal)
# Add a check to avoid multiplying if it's already a percentage (e.g., value > 1)
if obesity_data["Year6_%_obese"].max() < 2: # Heuristic: if max value is less than 2, it's likely a decimal (e.g. 0.15)
    obesity_data["Year6_%_obese"] = obesity_data["Year6_%_obese"] * 100

# Drop rows with missing Year 6 data
year6_data = obesity_data.dropna(subset=["Year6_%_obese"])

# Filter to include only 'Ward' entries
year6_data = year6_data[year6_data['Geog Level'] == 'Ward']

# Sort by highest obesity percentage
top_20_wards = year6_data.sort_values(
    by="Year6_%_obese",
    ascending=False
).head(20)

# Select key columns to display
display(top_20_wards[["Area name", "LA name", "Year6_%_obese"]])
```

```python
top_20_wards = year6_data.sort_values(
    by="Year6_%_obese",
    ascending=True
).head(20)

# Select key columns to display
display(top_20_wards[["Area name", "LA name", "Year6_%_obese"]])
```

next step could be calculating the average percentage of obesity for our boroughs with highest and lowest fast food density
so the average of all the wards within the boroughs
then plot to see correlation

```python
average_year6_by_borough = (
    obesity_data
    .groupby("LA name")["Year6_%_obese"]
    .mean()
    .reset_index()
    .rename(columns={"Year6_%_obese": "Avg_Year6_%_obese"})
)

# View results
average_year6_by_borough.head(32)
```

```python
average_year6_by_borough.sort_values(
    by="Avg_Year6_%_obese",
    ascending=False
).head()
```

```python
average_year6_by_borough.sort_values(
    by="Avg_Year6_%_obese",
    ascending=True
).head()
```

```python
import pandas as pd
from google.colab import files
import io

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
ethnicity_data = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name="Ethnic Group Boroughs") # Only selecting the sheet 'Ethnic Group Boroughs'
```

For each LA name, this dataset gives us the population of each ethnic group. Thus, the sum of all the populations equals to the total number of inhabitants of each LA.

```python
# let's clean column names by stripping whitespace
ethnicity_data.columns = [col.strip() for col in ethnicity_data.columns]

# let's rename the columns accordingly to the ones of Childhood Obesity and Count of Outlets
ethnicity_data = ethnicity_data.rename(columns={'Lower tier local authorities Code': 'LA code'})
ethnicity_data = ethnicity_data.rename(columns={'Lower tier local authorities': 'LA name'})


# let's remvoe any completely emplty columns that might have been created during the process
ethnicity_data = ethnicity_data.dropna(axis=1, how='all') #dropna used to remove missing values, axis=1 specifies this should apply to columns (axis=0, to rows)

ethnicity_data.head()
```

```python
# let's drop the 'Ethnic group (20 categories) Code' column as it's redundant with 'Ethnic group (20 categories)'
ethnicity_data = ethnicity_data.drop(columns=['Ethnic group (20 categories) Code'])

# let's convert 'Observation' to numeric, coercing errors to NaN
ethnicity_data['Observation'] = pd.to_numeric(ethnicity_data['Observation'], errors='coerce')

display(ethnicity_data.head())
display(ethnicity_data.info()) #check
```
